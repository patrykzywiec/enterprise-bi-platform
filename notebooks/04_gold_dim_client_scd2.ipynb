```python
# ============================================
# GOLD LAYER - DIM CLIENT
# ============================================

from pyspark.sql.functions import *

df_silver = spark.table("silver.transactions_clean")
```

```python
# --------------------------------------------
# Build client dimension with activity metrics
# --------------------------------------------

dim_client = (
    df_silver
    .groupBy("client_id")
    .agg(
        min("date").alias("first_transaction_date"),
        max("date").alias("last_transaction_date"),
        count("*").alias("total_transactions"),
        sum("amount").alias("lifetime_value")
    )
    .withColumn("client_sk", monotonically_increasing_id())
)
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")
```

```python
spark.sql("DROP TABLE IF EXISTS gold.dim_client")
dim_client.write.mode("overwrite").format("delta").saveAsTable("gold.dim_client")
```
